```python
match_wins=df['match_winner'].value_counts()
sys.barplot(y = match_wins.index,x = match_wins.values,palette = 'viridis')
plt.title("Most Match Winner")
```

```python
df['toss_decision'].value_counts()
```

```python
sys.countplot(x=df['toss_decision'],palette = 'viridis',hue = df['toss_decision'])
plt.title("Toss Decision")
```

```python
win=df[df['toss_winner']==df['match_winner']]['match_id'].count()
win = (win*100)/df.shape[0]
win = win.round(2)
loss=df[df['toss_winner']!=df['match_winner']]['match_id'].count()
loss = (loss*100)/df.shape[0]
loss = loss.round(2)

df1 = pd.DataFrame({
    'win' : [win],
    'loss' : [loss]
})

ax = sys.barplot(x=df1.columns,y=df1.iloc[0],palette='rainbow')
for container in ax.containers:
    ax.bar_label(container)
plt.xlabel("Status")
plt.ylabel("Percentage")
plt.title("Toss Winner VS Match Winner")
```

```python
sys.countplot(x=df['won_by'],palette='rainbow')
plt.title("How do Team Wins")
```

```python
High_Score = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
sys.barplot(x = High_Score.index,y = High_Score.values,palette='viridis')
plt.title("Top Scorer")
plt.xlabel("Player Name")
plt.ylabel("High Score")
```

```python
def extract_wicket(txt):
  wicket =""
  for i in txt:
    if i == '-':
      break
    else:
      wicket+=i
  return wicket
```

```python
df['Highest_Wicket'] = df['best_bowling_figure'].apply(extract_wicket)
df['Highest_Wicket'] = df['Highest_Wicket'].astype(int)
Bowling_Figure = df.groupby('best_bowling')['Highest_Wicket'].sum().sort_values(ascending=False).head(10)
sys.barplot(y=Bowling_Figure.index,x=Bowling_Figure.values,palette = 'viridis')
plt.ylabel("Bowler Name")
plt.xlabel("Highest Wicket")
plt.title("Top 10 Bowler with Highest Wicket")
```
